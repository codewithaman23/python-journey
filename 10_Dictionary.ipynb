## Shallow Copy --> Shallow copy is being used for creating a copy of the original dictionary in a way such that the changes that we are going to make with the original dictionary wont be getting reflected into our copied dictionary and shallow copy is done by copy()  method

```python
dictionary_1 = {'Name':'Aman', 'a':23, 'b':21, 'z':100}
```

```python
dictionary_2 = dictionary_1
```

```python
dictionary_2
```

```python
dictionary_1['a'] = 100
```

```python
dictionary_1
```

```python
dictionary_2 # chnanges also  happen in dictionary_2
```

```python
dictionary_1
```

```python
dictionary_3 = dictionary_1.copy()
```

```python
dictionary_3
```

```python
dictionary_1['a'] = 1000
```

# Nested Dictionary

Nested Dictionary will be used for keeping a dictionary inside another dictionary

```python
# want to store the information for multiple srtudnets like thire name ,maeks, grades

Nested_dictionary = {'Student_1': {'Name':'Aman','Marks':95, 'Grade': 'A'},
                     'Studnet_2': {'Name':'Anmol', 'Marks':90, 'Grade':'A'},
                     'Studnet_3': {'Name': 'Ansh', 'Marks': 85, 'Grade': 'A'} }
```

```python
Nested_dictionary
```

```python
# Iterate over these Nested Dictionary 
```

```python
for outer_key,outer_value in Nested_dictionary.keys:
    print(f'Outer key is : {outer_key} and outer value is : {outer_value}')
    for key_1, value_1 in outer_value.items():
        print(f"Inner key is: (key_1) and Inner value is : (Value_1)")
```

## Dictionary Comprehension

#### Dictionary comprehension helps us with looping statement along with conditional statement inside our dictionary

syntax of dictionary comprehension

{expression for item in iterable if condiiton}

```python
comprehensive={x**2 for x in range(11)}
```

```python
comprehensive
```

```python
comprehensive = {x:x**2 for x in range(11)}
```

```python
comprehensive
```
