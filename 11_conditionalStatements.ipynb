# Conditional Statements


Conditional Statements works like Real Life scenario where if a certain condition is meeting then a statement of code or a block of code gets executed.

```python
# Eligibility for driving on meeting a certain conditional

# Conditional Statements --> if statement where if a condition is meeting then a block of code gets executed.

Age = int(input('Enter your age value please :'))

if Age>=18:
    print('Yes you are eligible for driving');

else:
    print('You are not eligible for driving')
```

## Conditional Statements

1. if statements

2. elif statement

3. else statement

```python
## Figure out that who is the tallest among the two friends.

Aman_height = int(input('Enter that what your height is Aman:'))
Anmol_height = int(input('Enter that what your height is Anmol:'))

if Aman_height>Anmol_height:
    print('Aman is the taller than Anmol')

else:
    print('Anmol is taller then Aman')
```

```python
## Figure out that who is the tallest among the two friends.

Aman_height = int(input('Enter that what your height is Aman:'))
Anmol_height = int(input('Enter that what your height is Anmol:'))
Ansh_height = int(input('Enter that what your height is Ansh '))


if Aman_height>Anmol_height:
    print('Aman is the tallest among the three friends')

elif Anmol_height>Aman_height>Anmol_height:
    print('Anmol is the tallest among the three friends')

else:
    print('Ansh is the tallest among the three friends')
```

# Nested Conditional Statement

Nested Conditional Statements involve if statements inside another if statement

```python
# Compare the height of three friends

Aman_height = int(input("Enter Aman height: "))
Anmol_height = int(input("Enter Anmol height: "))
Ansh_height = int(input("Enter Ansh height: "))

if Aman_height > Anmol_height:
    if Aman_height > Ansh_height:
        print("Aman is the tallest guy")
    else:
        print("Ansh is the tallest guy")

elif Anmol_height > Ansh_height:
    print("Anmol is the tallest guy")

else:
    print("Ansh is the tallest guy")

```
