```python
# To crate a function that converts temperature from celcius to Fahrenheit or Faherenheit to Celcius

"""
formula --

Celcius -> Fahrenheit    F = (C * 9/5) + 32
Fahrenheit -> Celsius    C = (F - 32) * 5/9
"""

def temperature(temp,unit):
    if unit == 'C':
        fahrenheit = (temp * 9/5) + 32
        return fahrenheit

    elif unit == 'F':
        Celsius = (temp - 32) * 5/9
        return Celsius
    else:
        return 'Invalid Syntax'
```

```python
print(temperature(100,'C'))
print(temperature(212,'F'))
```

```python
## define a function that would behave as a Password Strength Checker
"""
Strong Password -->
if your password:
     1. length is greater then 8 characters.
     2. password must consist of atleat a lowercase character (a,b,c,...,z)
     3. password must consist of atleast a uppercase character (A,B,C,...,Z)
     4. password must consist of atleast a digit (1,2.)
     5. password must consist of atleast a special character (!,@,#,$,%,^...)

     Weak Password -->
     if a password does not match any above condition then it will consider as a weak password

"""
```

## islower() --> checks that wether if the current character is a lowercase character or not

## isupper() --> checks that wether if the current character is a uppercase character or not.

## isdigit() --> checks that wether if the current character is a digit or not



```python
def password_strength(password):
    if len(password) < 8:
        print(f"The entered password {password} is a Weak Password")

    elif not any(char.islower() for char in password):
        print(f"The entered password {password} is a Weak Password")

    elif not any(char.isupper() for char in password):
        print(f"The entered password {password} is a Weak Password")

    elif not any(char.isdigit() for char in password):
        print(f"The entered password {password} is a Weak Password")

    elif not any(char in "!@#$%^&*()_" for char in password):
        print(f"The entered password {password} is a Weak Password")
        
    else:
        print(f"The entered password {password} is a Strong Password")

    
     

    
```

```python
password_strength('AmanSingh23@5')
```

```python
def password(password):
    has_lower = False
    has_upper = False
    has_digit = False
    has_special = False

    for char in password:
        if char.islower():
            has_lower = True
        elif char.isupper():
            has_upper = True
        elif char.isdigit():
            has_digit = True
        elif char in '!@#$%^&*()_':
            has_special = True

    if len(password) < 8:
        print('Weak Password: password must have at least 8 characters')

    elif has_lower == False:
        print('Weak Password: lowercase letter missing')

    elif has_upper == False:
        print('Weak Password: uppercase letter missing')

    elif has_digit == False:
        print('Weak Password: digit is missing')

    elif has_special == False:
        print('Weak Password: special character is missing')

    else:
        print('Strong password')


        
```

Palindrome String

Palindrome String is going to have characters similar to it from both the sides or even if you will reverse the string still you will be getting the given string itself. 

reversing a string --> srtring[::-1]

```python
## check that whether if the current string is palindrome or not 

def palindrome(string):
    string_value = string.lower().replace(' ', '')

    if string_value == string_value[::-1]:
        print('The current string is a palindrome')

    else:
        print('The current string is not a palindrome')
   
```

```python
palindrome('Aba')
```

```python
palindrome('Ab')
```
