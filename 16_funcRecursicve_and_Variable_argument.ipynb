# Recursive Function

## By a Recursive Function we can say that a Function is going to call itself again and again

```python
# Factorial of a Number 
# Factorial of 8 -> 8*7*6*5*4*3*2*1
# Factorail of 0 or Factiorail of 1 --> 1
```

```python
def factorial(num):
    if num == 0 or num == 1:
        print(f'The Factorial of {num} is : 1')

    else:
        factorial = 1
        for i in range(1, num + 1):
            factorial = factorial * i
        print(f'The factorial of a num is {factorial}')
```

```python
number = int(input('Enter the number:'))
factorial(number)
```

```python
def factotial_recursive(num):
    if num == 0 or num == 1:
        return 1

    else:
        return num * factotial_recursive(num - 1)
```

```python
factotial_recursive(5)
```

```python
factotial_recursive(2)
```

## Functions with Multiple parameters

In functions with multiple parameters , we are going to pass multiple values as our parameter or more then one parameter we can pass here.

## Function with Variables length argumnets

## Positional Arguments --> Positional argumnets are being used so that we can pass multiple values inside our function as an argument irrespective of the number of values we have mentioned as the parameter inside our function.

```python
## Function to add multiple numbers irrespective of how many numbers we are passing

def addition_position(*num):
    add = 0
    for i in num:
        add = add + i

    print('The addition of the numbers are', add)
```

```python
addition_position(4,4,4)
```

## Keyword Arguments --> Takes the input as the keyword values and irrespective of any number of keyword values you are going to pass it wont matter

```python
def data(**user_data):
    for key,value in user_data.items():
        print(f'{key}:{value}')
```

```python
data(name='aman', age= 25, gender ='male')
```

## You guys want to pass keyword value pairs as your input along with positional arguments as well

```python
def variable(*positional, **keyword):
    for i in positional:
        print(f'positional values are:',i)

    for key,value in keyword.items():
        print(f'key is :{key} and value is : {value}')
```

```python
variable(1,2,3,4,name='aman',age = 24,Qualification = 'Btech in Computer Science')
```
