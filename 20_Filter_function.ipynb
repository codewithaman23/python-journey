# Filter Function

## Filter Function is going to retieve only those items which are meeting a certain condition in an iterable

#### The filter function constructs an Iterator from elements of an iterable for which a condition remains True or for which a Function returns True Value

#### Syntax for filter Function

list(filter(func, iterable))

```python
## To retieve only those values which are even

def even_function(value):
    if value % 2 == 0:
        return True
```

```python
iterable_list = [1,4,5,9,2,7,11,24,29,28,32,42,67,72,85,96,100]
print(list(filter(even_function, iterable_list )))
```

```python
## solving the similar problem statement using lambda function

print(list(filter(lambda x:x%2 == 0 , iterable_list)))
```

```python
## I want to retrive the value which are greater than 5

print(list(filter(lambda x:x>5,iterable_list)))
```

```python
## I want to retrieve al those values which are greater than 5 as well and are even as well

list_items = [1,7,8,10,17,20,24,47,68,72,100]
print(list(filter(lambda x:x>5 and x%2==0,list_items)))

```

```python
## Iterating over the dictionary items

employees = [{'Name':'Aman','Age':25,'Qualification':'Btech'},
             {'Name':'Anmol','Age':26,'Qualification':'Btech'},
             {'Name':'Ansh','Age':27,'Qualification':'Btech'},
             {'Name':'Rohit','Age':28,'Qualification':'Btech'}]
```

```python
def age_greater(employees):
    if employees['Age']>25:
        return True
```

```python
print(list(filter(age_greater, employees)))
```
