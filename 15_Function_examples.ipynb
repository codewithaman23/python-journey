## Functions are reusable and by that we mean to say is We dont have to write the same piece of code again and again and this is the sign of a good programmer as well as we dont have to write the same lines of code again and along with that it can help us out organizing the code as well.

```python
# you are interviewer who is welcoming the candidate everytime with same greeting so ypu plan to use functions to avoid similar greeting again and again.

def greeting(candidate_name):
    """ This function takes the input as the user name and provide the output as the greeting"""
    print(f"Hello {candidate_name}, How are you? We really liked your previous experience and your skillset and now we are planning to move forward with you but before that if you can let us know more about you.")
```

```python
greeting('Aman')
```

```python
greeting('Neha')
```

## Why Functions 

### It increases Readability, Reusability and for organizing the code as well.


```python
# Creating a function which is going to take input number from the user side and will figure out that whether if the number is even or odd

def even_or_odd(num):
    if num % 2 == 0:
        output = 'Even number'
    else:
        output = 'Odd number'

    return output
    
```

```python
print(even_or_odd(2))
```

```python
print(even_or_odd(996))
```

## Handling Errors in out function based in what input the user is going to provide to the function

```python
print(even_or_odd(s))
```

Ask the user to provide with a proper input or if you are a good programmer then you wont be getting into this hassle of correcting the user and you can handle the issue on your own

```python
def even_or_odd_instance(num):
    if isinstance(num, int):
        if num%2==0:
            print('Even number')
        else:
            print('Odd number')

    elif isinstance(num,str):
        print('Invalid data type')
    else:
        print(f'{num} is an invalid value for this function')
```

```python
even_or_odd_instance(2)
```

```python
even_or_odd_instance('s')
```

```python
even_or_odd_instance(True)
```

```python
even_or_odd_instance(14.4)
```
