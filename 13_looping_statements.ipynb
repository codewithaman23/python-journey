# while Loop

while loop is being used for iterating over a series of numbers when a condition is True and we have to initialise a value as well while iterating 

```python
# by making the use of while loop we need to print the numbers from 0 to 50

i = 0
while i<=50:
    print(i)
    i = i + 1
```

```python
# print the number from 0 to 50 as the even numbers

for i in range(51):
    if(i%2==0):
        print(i)
```

# Loop Control Statements

Loop Control Statements are being used for terminating a loop in between when a condition is met or if you want to have some control over your iterating process or while iterating over the series of numbers.

### break --> break is our first loop control statement which will help you out to terminate the loop when a certain condiiton is met

```python
# print the number from 0 to 50 but once number = 18 then terminate the loop or do not print either 18 or any number forward

for i in range(51):
    if i == 18:
        break
    else:
        print(i)
```

## continue --> continue is are second loop control statement which helps us in iterating over a series of numbers but helps us with skipping the values in between once we meet a condition

```python
## print the number from 0 to 50 but only print even numbers and for odd numbers skips those values 

for i in range(51):
    if i % 2 == 0:
        print(i)
    else:
        continue
```

```python
# print the number from 0 to 20 but once 17 is occuring skip it 

for i in range(21):
    if i == 17:
        continue
    else:
        print(i)
    

```

## Pass --> Pass statement is being used for doing nothing

```python
# print the number from 0 to 20 but once 17 occuring do not 

for i in range(21):
    if i == 17:
        print('Our program will do nothing since the value is 17 and it wont be getting printed')
        pass
    else:
        print(i)
```

# Nested Loop

Neseted Loops --> it is going to have a loop inside a loop

Nested loops can be help us out with iterating over a series of numbers as with the proper combination of iterating numbers 

f string --> f string will be helping us out with printing the numbers with the variable value present inside the colons so we can directly represent the varaible value inside the colons message itself where variable value will be shown inside the {}

```python
for i in range(6):
    for j in range(4):
        print(f"i value is: {i} and j value is: {j}")
```

```python
# printing the sum of First N Natural numbers where Natural number --> 1,2,3.....n

n_value = int(input("Enter the value for n:"))
sum = 0
for i in range (n_value+1):
    sum+=i
print(f"The sum of first {n_value} natural numbers is : {sum}")
```
