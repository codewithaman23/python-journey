# Building a simple calculator

```python
value_1 = int(input('Enter your first value:'))
value_2 = int(input('Enter your second value:'))
print('The addition for the value is:', value_1 + value_2)
print('The subtraction for the value is:', value_1 - value_2)
print('The product for the value is:', value_1 * value_2)
print('The division for the value is:', value_1 / value_2)
```
