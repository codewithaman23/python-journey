# Types of Operator

1. Arithmetic Operator
Addition (+)
Substraction (-)
Product or Multiplicaation (*)
Division (/)
Modulus (%) -> it gives the remaninder
Floor (//)
Exponent (a ** b)

2. Comparison Operator -> It checks the two given operatends or values are equal or not

Greater than (>)
Less than (<)
Greater than or equal (>=)
Less than or equal (<=)
Equality check (==)
Not equal to (!=)


3. Logical Operators

and --> whether if both condition are true or not
or --> whether if any of the operation is true or not
not --> negation or to inverse the result of the given opeation.

```python
# simple calculator

a= int(input("Enter the first value"))
b= int(input("Enter the second value"))
operation = input("Enter the operation out of +,-,*,/")

if operation == "+":
    print("The addition of the two values is:", a+b)

if operation == "-":
    print("The subtraction of the two values is:", a-b)

if operation == "*":
    print("The product of the two values is:", a*b)

if operation == "/":
    print("The division of the two values is:", a/b)
```
