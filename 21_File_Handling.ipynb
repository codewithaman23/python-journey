# File Handling 

## 1. Read the content from My Text File in the system

## open() --> opening the file
## First Parameter --> File_name
## Second Parameter --> Mode
## Mode --> It decides the way in which we are working with this file

#### Mode --> r --> Reading the content from the File.
#### Mode --> w --> Write the content into the File.
#### Mode --> r+ --> Read the content first and then writing the content into the File.
#### Mode --> w+ --> Writing the content first and then reading from it.



## Reading your file

read() --> Reads the conetent from the file

```python
f = open('example.txt', 'r')
content = f.read()
print(content)
f.close()
```

## Writing the Content into the file

## write() --> it overwrite the content into the text files.
## append() --> it adds the data from the end of the file

```python
## Write() --> helps us ou tin wwriting the content into the file

f = open('example.txt', 'w')
f.write('Hello we are now writing the content into our file')
f.close()
```

```python
## Appending the content into our File

f = open('example.txt','a')
f.write('\n We are learning about Gen Ai.')
f.close()
```

## with --> It will be used for first of all opening the file as a object and there is no requirement for closing the file as well


```python
with open('example.txt','r') as f:
    content_data = f.read()
    print(content_data)
```

```python
with open('example.txt','w') as file:
    file.write('We are exploring about with command in python File Handling')
```

```python
with open('example.txt','a') as files:
    files.write('Hello we are exploring append mode as well')
```
