# Lambda Functions

## Lambda Functions are the One Liner Functions defined by the Lambda keyword

## Syntax for Lambda Function:

lambda arguments:expression

lambda --> keyword which simpally let the python interpreter know that the current function is our Lambda Function

Arguments --> Input values for our function

Expression --> Logic (Function Body)

## lambda x:x**2

## They can have any number of Arguments but only one expression or only one Logic

## They are used for Short Operations or as Arguments to Higher Order Function

## It can be treated as a Function without a Name and they Dont even require a Return Statement

```python
## Function which calculates the square of the numbers

def square(num):
    return num**2
```

```python
square(10)
```

```python
## This similar problem can be solved using Lambda Functions as well

lambda_function=lambda x:x**2
```

```python
type(lambda_function)
```

```python
lambda_function(5)
```

```python
## Lambda function to calculate the sum of two numbers 

sum_variable=lambda x,y:x+y
```

```python
sum_variable(10,10)
```

```python
product = lambda a,x,y,z:a*x*y*z
print(product(10,10,10,10))
```
