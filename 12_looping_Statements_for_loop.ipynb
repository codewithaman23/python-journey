# Looping Statements

Looping Statements can be used for iterating over a series of numbers or through iterables or Data Structures or through a collection of data.

# range()

range() can be used for working as our iterable 

range() takes an input parameter and the first parameter that the range will be taking as the input will be the initial value that you want to print or you want to operate on and then you will be passing the next input after comma as the parameter as the final value and it can take third parameter as well and that will be the step size and that is the difference between the iterating values.

#### range (initial_value , final_value + 1, step_size)

```python
# print the number from 1 to 50

for item in range (1,51,1):
    print(item)
```

In the range function there is never a necessary for mentioning the step size and similarly if you want to print the numbers or you want to operate on numbers again there is no neccessity for passing the initial value as well if and only if your initial value is 0.

```python
# numbers from 0 to 100

for i in range(101):
    print(i)
```

```python
# to print the number from 0 to 100 with a gap of 5

for i in range(0,101,5):
    print(i)
```

By using range() we can represent the number in a decreamenting manner as well or in a decreasing order as well.

```python
# print the number from 50 to 1 with a gap of 1 value in a decremenitng manner

for i in range(50,0,-1):
    print(i)
```

```python
# print the number from 50 to 0 with a gap of 5

for i in range(50,-1,-5):
    print(i)
```

```python
# print the number from 500 to 0 with a gap of 5

for i in range(500,-1,-5):
    print(i)
```
